```python
bd=pd.read_csv('owid-covid-data.csv')
```

```python
fecha = '2021-05-18'
```

```python
pais = 'Mexico'
```

```python
bd['date'] = pd.to_datetime(bd['date'])
#filtro para México
mexico_data = bd[(bd['location'] == pais) & (bd['date'] == fecha)]
```

```python
try:
    # verificar datos
    if not mexico_data.empty:
        # total de casos para esa fecha en México.
        total_cases_mexico = mexico_data['total_cases'].iloc[0]
        # Imprime los datos
        print("Pais         fecha            total cases")
        print(f"Mexico            {fecha}        {int(total_cases_mexico)}")
    else:
        print(f"No se encontraron datos para {pais} en la fecha {fecha}.")

except FileNotFoundError:
    print(f"Error: The file was not found. Make sure the file exists and the path is correct.")
except KeyError as e:
    print(f"Error: An expected column was not found in the CSV. Make sure the CSV has the columns 'location', 'date' and 'total_cases'. Detail: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
