```python
print("\n--- Top 5 juegos más vendidos desde el año 2000 ---")
```

```python
# 1. Filtrar los juegos desde el año 2000

bd['Release.Year'] = pd.to_numeric(bd['Release.Year'], errors='coerce')
bd_filtrado = bd[bd['Release.Year'] >= 2000].dropna(subset=['Release.Year'])
```

```python
# 2. Ordenar por ventas y mostrar el top 5
top_5_ventas = bd_filtrado.sort_values(by='Metrics.Sales', ascending=False).head(5)
display(top_5_ventas)
```
