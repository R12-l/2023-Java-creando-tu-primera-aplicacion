```python
# Importar librerías

import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img_cv = cv2.imread('/content/drive/MyDrive/APTOS-5C/val/0/00cc2b75cddd.png')
```

```python
print(type(img_cv)) # Tipo de dato
print(" ")
print(img_cv.dtype) # Tipo de dato de los elementos (0 - 255)
```

```python
plt.imshow(img_cv)
plt.show()
# OpenCV lee la imagen en BGR pero Matplotlib espera una imagen en formato RGB
```

```python
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.show()
```

```python
# Conversión de imagen (color) a imagen a escala de grises
img_gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap = 'gray')
plt.show()
```

```python
# Leer y convetir la imagen (color) a escala de grises en una sola linea de código
img_gray_2 = cv2.imread('/content/drive/MyDrive/APTOS-5C/val/0/00cc2b75cddd.png', cv2.IMREAD_GRAYSCALE)
plt.imshow(img_gray_2, cmap = 'gray')
plt.show()
```

```python
img_rsize = cv2.resize(img_rgb, (500, 350)) #Nuevo ancho, nuevo alto
```

```python
plt.imshow(img_rsize)
plt.show()
```

```python
img_rsize_2 = cv2.resize(img_rgb, (100, 175))
plt.imshow(img_rsize_2)
plt.show()
```

```python
# Redimensionar manteniendo la proporción de la imagen
nvo_ancho = int(img_rgb.shape[1]*0.5)
nvo_alto = int(img_rgb.shape[0]*0.5)
img_rsize_3 = cv2.resize(img_rgb, (nvo_ancho, nvo_alto))

# (350, 500, 3) = altura 350, ancho 500, 3
```

```python
# Ancho = 500 * 0.5 = 250 píxeles
# Alto = 350 *0.5 = 175 píxeles
```

```python
plt.imshow(img_rsize_3)
plt.show()
```

```python
# Flipping X, Y, XY

img_flipX = cv2.flip(img_rgb, 0)
plt.imshow(img_flipX)
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10,5))

# Imagen original
axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Orginal')
axs[0].axis('off')

# Imagen volteada
axs[1].imshow(img_flipX)
axs[1].set_title('Imagen volteada (Flip X)')
axs[1].axis('off')

# Mostrar ambas imágenes
plt.show()
```

```python
img_flipY = cv2.flip(img_rgb, 1)
plt.imshow(img_flipY)
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10,5))

# Imagen original
axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Orginal')
axs[0].axis('off')

# Imagen volteada
axs[1].imshow(img_flipY)
axs[1].set_title('Imagen volteada (Flip Y)')
axs[1].axis('off')

# Mostrar ambas imágenes
plt.show()
```

```python
img_flipXY = cv2.flip(img_rgb, -1)
plt.imshow(img_flipXY)
plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.suptitle("Comparación de transformaciones de Imagen")
plt.subplot(2,2,1)
plt.imshow(img_rgb)
plt.title("Imagen Original")
plt.subplot(2,2,2)
plt.imshow(img_flipX)
plt.title("Flipped along x-axis")
plt.subplot(2,2,3)
plt.imshow(img_flipY)
plt.title("Flipped along y-axis")
plt.subplot(2,2,4)
plt.imshow(img_flipXY)
plt.title("Flipped along xy-axis")
plt.show()
```

```python
feto = cv2.imread('/content/drive/MyDrive/ultrasonido.tif')
plt.imshow(feto)
plt.show()
```

```python
img_rgb_feto = cv2.cvtColor(feto, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb_feto)
plt.show()
```

```python
img_flipX_feto = cv2.flip(img_rgb_feto, 0)
img_flipY_feto = cv2.flip(img_rgb_feto, 1)
img_flipXY_feto = cv2.flip(img_rgb_feto, -1)
```

```python
plt.figure(figsize=(12,8))
plt.suptitle("Comparación de transformaciones de imagen - feto")
plt.subplot(2,2,1)
plt.imshow(img_rgb_feto)
plt.title("Imagen Original")
plt.subplot(2,2,2)
plt.imshow(img_flipX_feto)
plt.title("Flipped along x-axis")
plt.subplot(2,2,3)
plt.imshow(img_flipY_feto)
plt.title("Flipped along y-axis")
plt.subplot(2,2,4)
plt.imshow(img_flipXY_feto)
plt.title("Flipped along xy-axis")
plt.show()
```
