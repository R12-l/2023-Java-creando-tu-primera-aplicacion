```python
img = cv2.imread('/content/drive/MyDrive/angel.jpg')
```

```python
print(type(img))
print("")
print(img.dtype)
print("")
print(img.shape)
```

```python
plt.figure(figsize=(12,6))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```

```python
heigth, width, channels = img.shape
print(heigth)
print(width)
print(channels)
```

```python
Ty= heigth/4 #desplazameinto vertical donde se desplazara hacia abajo en una 1/4 parte de la altura de la iamgen
Tx= width/4 # desplazamiento horizontal donde se desplazara hacia la derecha en una 1/4 parte ancho de la imagen
```

```python
#matriz de translacion
T = np.float32([[1,0,Tx],[0,1,Ty]])
print(T)
```

```python
img_translation = cv2.warpAffine(img, T, (width, heigth))
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Imagen original')
plt.subplot(1,2,2)
plt.imshow(img_translation)
plt.show()
```

```python
Ty= -50
Tx=-30
T = np.float32([[1,0,Tx],[0,1,Ty]])
print(T)
```

```python
img_translation = cv2.warpAffine(img, T, (width, heigth))
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Imagen original')
plt.subplot(1,2,2)
plt.imshow(img_translation)
plt
```

```python
#rotacion
R=cv2.getRotationMatrix2D(center=(width/2, heigth/2), angle=90, scale=1)
print(R)
```

```python
img_rotated = cv2.warpAffine(img, R, dsize=(width, heigth))
```

```python

plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Imagen original')
plt.subplot(1,2,2)
plt.imshow(img_rotated)
plt
```
