```python
# --- Inventario de la Tienda ---
print("--- Inventario de la Tienda ---")

# Creamos un diccionario vacío para guardar los ítems y sus precios.
# Un diccionario es como una lista de pares: "nombre del ítem" y "precio".
inventario = {}

# Pedimos al usuario que ingrese los ítems y sus precios.
# El bucle seguirá pidiendo ítems hasta que el usuario escriba 'listo'.
while True:
    prod = input("Ingresa el nombre del producto (o 'listo' para terminar): ").lower()
    if prod == 'listo':
        break
    try:
        precio = float(input(f"Ingresa el precio de '{prod}': "))
        inventario[prod] = precio
    except ValueError:
        print("Por favor, ingresa un número válido para el precio.")

# Mostramos el inventario que el usuario acaba de crear.
print("\nInventario actual:")
for prod, precio in inventario.items():
    print(f"- {item.capitalize()}: ${precio:.2f}")

# --- Ingredientes de la Receta ---
print("\n--- Ingredientes de la Receta ---")

# Creamos una lista vacía para guardar los ingredientes de la receta.
# Una lista es como una secuencia de elementos.
ingredientes_receta = []

# Pedimos al usuario que ingrese los ingredientes de la receta.
# El bucle seguirá pidiendo ingredientes hasta que el usuario escriba 'listo'.
while True:
    ingrediente = input("Ingresa un ingrediente para la receta (o 'listo' para terminar): ").lower()
    if ingrediente == 'listo':
        break
    ingredientes_receta.append(ingrediente)

# Convertimos la lista de ingredientes a un conjunto.
# Un conjunto automáticamente elimina los ingredientes repetidos,
# así solo tenemos los ingredientes únicos.
ingredientes_unicos = set(ingredientes_receta)

# Mostramos los ingredientes únicos para la poción.
print(f"\nIngredientes únicos para la poción: {ingredientes_unicos}")
```
