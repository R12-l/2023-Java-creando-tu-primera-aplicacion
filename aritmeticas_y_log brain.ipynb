```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
#realizar la suma
im1=cv2.imread('g1.jpg',0) #el 0 lee la iamgen en monocromatico

```

```python
print(type(im1))
print("")
print(im1.dtype)
print("")
print(im1.shape)
```

```python
plt.imshow(im1,cmap='gray')
plt.show()
```

```python
print(im1)
```

```python
#podemos suamr escalar o imagen con imagen
#crearemos una iamgene scalar con el msimo tamaño de la iamgen 1
scalar=np.ones(im1.shape, dtype='uint8')*150 #crea una matriz array en 1
print(scalar)
```

```python
add_im1=cv2.add(im1,scalar)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(im1, cmap = 'gray')
plt.title("Imagen Original")
plt.subplot(1,2,2)
plt.imshow(add_im1, cmap = 'gray')
plt.title("Imagen - Suma")
plt.show()
```

```python
sub_im1=cv2.subtract(im1,scalar)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(im1, cmap = 'gray')
plt.title("Imagen Original")
plt.subplot(1,2,2)
plt.imshow(sub_im1, cmap = 'gray')
plt.title("Imagen - resta")
plt.show()
```

```python
#multiplicacion
mul_im1=cv2.multiply(im1,4)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(im1, cmap = 'gray')
plt.title("Imagen Original")
plt.subplot(1,2,2)
plt.imshow(mul_im1, cmap = 'gray')
plt.title("Imagen - multiplicaciòn")
plt.show()
```

```python
#division
div_im1=cv2.divide(im1,20) #cada pixel de la iamgen se divide entre 20
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(im1, cmap = 'gray')
plt.title("Imagen Original")
plt.subplot(1,2,2)
plt.imshow(div_im1, cmap = 'gray')
plt.title("Imagen - division")
plt.show()
```

```python
#operaciones logicas
im2=cv2.imread('black.png',0)
```

```python
print(type(im2))
print("")
print(im2.dtype)
print("")
print(im2.shape)
```

```python
im1=cv2.resize(im1,(512,512))
print(im1.shape)
```

```python
plt.imshow(im1,cmap='gray')
plt.title("$imagen_1$")
plt.show()
```

```python
im2=cv2.imread('black.png',0)
im2=cv2.resize(im2,(512,512))
print(im2.shape)
```

```python
plt.imshow(im2,cmap='gray')
plt.title("$imagen_2$")
plt.show()
```

```python
#logica AND
and_image=cv2.bitwise_and(im1,im2)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.imshow(im1, cmap = 'gray')
plt.title("$imagen1$")
plt.subplot(1,3,2)
plt.imshow(im2, cmap = 'gray')
plt.title("$imagen2$")
plt.subplot(1,3,3)
plt.imshow(and_image, cmap = 'gray')
plt.title("$imagen1$ AND $imagen2$")
plt.show
```

```python
#logical or
or_image=cv2.bitwise_or(im1,im2)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.imshow(im1, cmap = 'gray')
plt.title("$X_1$")
plt.subplot(1,3,2)
plt.imshow(im2, cmap = 'gray')
plt.title("$X_2$")
plt.subplot(1,3,3)
plt.imshow(or_image, cmap = 'gray')
plt.title("$X_1$ OR $X_2$")
plt.show
```

```python
#logical XPR
xor_image=cv2.bitwise_xor(im1,im2)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.imshow(im1, cmap = 'gray')
plt.title("$imagen_1$")
plt.subplot(1,3,2)
plt.imshow(im2, cmap = 'gray')
plt.title("$imagen_2$")
plt.subplot(1,3,3)
plt.imshow(xor_image, cmap = 'gray')
plt.title("$imagen_1$ XOR $imagen_2$")
plt.show
```

```python
#logical not
not_X1=cv2.bitwise_not(im1)
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.imshow(im1, cmap = 'gray')
plt.title("$imagen_1$")
plt.subplot(1,3,3)
plt.imshow(not_X1, cmap = 'gray')
plt.title("$iamgen_1$ NOT $imagen_2$")
plt.show
```
